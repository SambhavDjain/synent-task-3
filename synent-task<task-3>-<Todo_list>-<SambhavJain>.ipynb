###The To-Do List (CLI):
The To-Do List (CLI) is a simple command-line task manager that allows users to organize and manage their daily tasks. It provides a menu-driven interface where users can add new tasks, view all existing tasks, and delete tasks as needed.
Tasks are stored in a list during program execution and are displayed with numbering for easy selection. The program includes basic input validation to handle incorrect or invalid entries, ensuring smooth operation.
This project demonstrates fundamental programming concepts such as lists, loops, conditional statements, and user interaction in a command-line environment.

```python
tasks = []

def show_menu():
    print("\n--- Task Manager ---")
    print("A - Add Task")
    print("V - View Tasks")
    print("D - Delete Task")
    print("Q - Quit")


def view():
    print("\nTasks:")
    if len(tasks) == 0:
        print("No tasks yet.")
    else:
        count = 1
        for t in tasks:
            print(f"{count}) {t}")
            count += 1


def add():
    data = input("New task: ")
    if data.strip() == "":
        print("Cannot add empty task.")
    else:
        tasks.append(data)
        print("Added.")


def delete():
    if not tasks:
        print("Nothing to delete.")
        return

    view()
    choice = input("Delete task number: ")

    if not choice.isnumeric():
        print("Invalid input.")
        return

    idx = int(choice) - 1

    if idx < 0 or idx >= len(tasks):
        print("Out of range.")
        return

    print("Deleted:", tasks[idx])
    del tasks[idx]


while True:
    show_menu()
    option = input("Choose option: ").upper()

    match option:
        case 'A':
            add()
        case 'V':
            view()
        case 'D':
            delete()
        case 'Q':
            print("Goodbye!")
            break
        case _:
            print("Invalid option.")
```
